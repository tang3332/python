```python
import math

#定义简单函数
def say_hi():
    print("Hello,world!")

def goat_pad(s):
    print("goat " + s + " goat")

def print_n(s,n):
    for _ in range(n):
        print(s)

say_hi()
goat_pad("hello,world")
print_n("python",6)
```

```python
#欧几里得算法
def gcd(m,n):
    #最大公约数一定比两个数都小，所以我采用了暴力求解
    i = min(m,n)
    while i>=1 :
        if m % i == 0 and n % i == 0:
            return i
        i-=1

print(gcd(2018,2019))
print(gcd(1200,300))
print(gcd(5040,60))
#如果输入是负数，我的代码求不出答案，没有意义
print(gcd(-5040,120))
print(gcd(-1200,-30))
```

```python
#近似欧拉数e
def euler_limit(n):
    return (1 + 1/n) ** n

def euler_infinite_sum(n):
    sum = 0.0
    #t是阶乘值
    t = 1
    for k in range(n):
        #应该从1开始求和，默认k=0时返回1
        if k > 0:
            #求阶乘
            t = t * k
        sum += 1.0/t
    return sum

def euler_approx(ep):
    #e为真实值
    e = math.exp(1)
    sum = 0.0
    t = 1
    k = 0
    while True:
        m = 1.0/t
        sum += m
        #求到在该值内接近真实值则停止循环
        if abs(e - sum)<ep:
            break
        k += 1
        t = t * k
    return sum

def print_euler_sum_table(n):
    for k in range(1,n+1):
        m = euler_infinite_sum(k)
        print(m)

print_euler_sum_table(5)
print(euler_infinite_sum(1))
```

```python
#测试整数属性
def is_power_of_2(n):
    #考虑到特殊情况，1可以是所以数的幂
    if n == 1:
        return True
    #如果n是2的倍数，则进入循环，减少运算，同时完成计算时退出
    while n % 2 == 0:
        n = n // 2
    #如果最后n的值为1，则说明是2的幂，返回布尔值
    return n == 1

def is_power(b,n):
    if n == 1:
        return True
    #因为1是1的幂，但是下面的循环没有考虑
    elif b == 1:
        return False
    while n % b == 0 and b > 1:
        n = n // b
    return n == 1
```
