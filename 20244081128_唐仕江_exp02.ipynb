```python
# bonjour
#NumPy数组、数据类型与索引
import numpy as np
import pandas as pd

#任务1
a1 = np.array([1, 2, 3])
a2 = np.array([1.5, 2.5, 3.5])
a3 = np.array([True, False, True])
print(a1.dtype, a1.shape, a1.ndim)
print(a2.dtype, a2.shape, a2.ndim)
print(a3.dtype, a3.shape, a3.ndim)
#ndarray所有元素是同一种类型，Python列表可以混着放不同类型，数组算得快
print()

#任务2
b = np.arange(12)
b1 = b.reshape(3, 4)
b2 = b.reshape(2, 3, 2)
print(b1)
print(b2)
print(b.size, 3 * 4, 2 * 3 * 2)
#改变形状时各维度相乘必须等于元素总数，不然报错
print()

#任务3
print(b1[1], b1[1].shape)
print(b1[:, 0:3])
print(b1[b1 > 6], b1[b1 > 6].shape)
#布尔索引筛出来的结果是一维的
print()

#任务4
c = np.arange(10)
c1 = c[(c < 2) | (c > 7)]
c2 = c[np.logical_or(c < 2, c > 7)]
c3 = c[(c > 3) & (c < 7)]
c4 = c[np.logical_and(c > 3, c < 7)]
print(c1, c2)
print(c3, c4)
#每个条件要加括号，|和&的优先级比比较运算符高
print()

#任务5
d1 = np.array([[1, 2], [3, 4]])
d2 = np.array([[5, 6], [7, 8]])
print(d1 * d2)
print(np.multiply(d1, d2))
print(d1 @ d2)
print(np.matmul(d1, d2))
#*是对应位置相乘，@是矩阵乘法，结果不一样
print()
```

```python
#随机抽样、统计与缺失值
#固定随机种子，保证每次运行结果一致
np.random.seed(2024)

#任务1
s1 = np.random.normal(70, 10, 1000)
s2 = np.clip(s1, 0, 100)
print(s2.size)
print(s2.min(), s2.max())
print(s2.mean(), s2.std())
print(s2.max() - s2.min())
print()

#任务2
n = np.arange(1, 11)
n1 = np.random.choice(n, 5, replace=False)
n2 = np.random.choice(n, 5, replace=False, p=n / n.sum())
print(n1)
print(n2)
#第一个是均匀抽样，第二个按数的大小加权，数越大概率越大
print()

#任务3
m = np.arange(1, 6)
m1 = np.random.permutation(m)
print(m, m1)
#permutation返回新数组，原数组不变
m2 = np.arange(1, 6)
np.random.shuffle(m2)
print(m2)
#shuffle直接原地打乱原数组
print()

#任务4
s3 = s2.copy()
s3[9] = np.nan
s3[19] = np.nan
s3[29] = np.nan
print(s3.mean())
print(np.nanmean(s3))
print(np.isnan(s3).sum())
#普通mean遇到nan结果就是nan，要用nanmean忽略缺失值
print()

#任务5
#含缺失值的成绩整理成表格变量，供实验4用，不写磁盘
table = pd.DataFrame({"序号": range(1, len(s3) + 1), "成绩": s3})
print(table.head())
print()
```

```python
#SciPy数值计算与Matplotlib可视化
from scipy import linalg, optimize, stats
import matplotlib.pyplot as plt

#任务1
#解方程组3x+y=9，x+2y=8
A = np.array([[3, 1], [1, 2]])
b = np.array([9, 8])
x = linalg.solve(A, b)
print(x)
print(A @ x)
#A@x等于b，说明解是对的
print()

#任务2
#求x²-2=0的根，初值一个给正的一个给负的
def f(x):
    return x ** 2 - 2

r1 = optimize.root(f, 1.0)
r2 = optimize.root(f, -1.0)
print(r1.success, r1.x, f(r1.x))
print(r2.success, r2.x, f(r2.x))
#两个根是±√2，函数值接近0说明求根成功
print()

#任务3
#求f(x)=x²+x+2的最小值
def g(x):
    return x ** 2 + x + 2

res = optimize.minimize(g, 0)
print(res.x, res.fun)
print(g(-0.5))
#解析解是x=-0.5，最小值1.75，和数值解一致
print()

#任务4
#两组独立样本做t检验
np.random.seed(2024)
x1 = np.random.normal(0, 1, 100)
x2 = np.random.normal(1, 1, 100)
t, p = stats.ttest_ind(x1, x2)
print(t, p)
#p小于0.05，两组均值差异显著
print()

#任务5
#有效成绩就是去掉nan后的成绩
scores_ok = s2[~np.isnan(s2)]

plt.hist(scores_ok, bins=30)
plt.title("有效成绩直方图")
plt.xlabel("成绩")
plt.ylabel("频数")
plt.show()

plt.boxplot(scores_ok)
plt.title("有效成绩箱线图")
plt.ylabel("成绩")
plt.show()

x = np.linspace(0, 5, 100)
plt.plot(x, x, label="y=x")
plt.plot(x, x ** 2, label="y=x²")
plt.plot(x, x ** 3, label="y=x³")
plt.title("y=x、y=x²、y=x³")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
#pandas的Series、DataFrame与内存中的CSV读写
#任务1
data = {
    "Name": ["Alice", "Bob", "Carol", "David", "Eva", "Frank"],
    "Group": ["A", "A", "B", "B", "A", "B"],
    "Math": [88, 72, 91, 65, 79, 84],
    "English": [90, 68, 85, 73, 92, 77],
    "Python": [95, 80, 89, 70, 86, 82],
}
df = pd.DataFrame(data)
df = df.set_index("Name")
df["Total"] = df[["Math", "English", "Python"]].sum(axis=1)
df["Average"] = df[["Math", "English", "Python"]].mean(axis=1)
print(df.index)
print(df.columns)
print(df.head(3))
print(df.describe().round(2))
print()

#任务2
print(df.loc["Carol"])
print(df.iloc[2])
print(df[df["Average"] >= 80])
print()

#任务3
print(df[["Math", "English", "Python"]].mean(axis=0))
print(df[["Math", "English", "Python"]].mean(axis=1))
print()

#任务4
df["Best"] = df[["Math", "English", "Python"]].apply(max, axis=1)
print(df["Best"])
print(df[["Math", "English", "Python"]].agg(["min", "mean", "max"]))
df["MathRank"] = df["Math"].rank(ascending=False)
print(df["MathRank"])
print()

#任务5
print(df.groupby("Group").size())
print(df.groupby("Group")[["Math", "English", "Python"]].mean().round(2))
#transform把组平均分放回每一行，行数不变
df["GroupPythonMean"] = df.groupby("Group")["Python"].transform("mean")
print(df[["Group", "Python", "GroupPythonMean"]].round(2))
print()

#任务6
#filter保留Python平均分不低于80的组
df2 = df.groupby("Group").filter(lambda g: g["Python"].mean() >= 80)
print(df2["Group"].unique())
print(df2.index)
print()

#任务7
print(df[["Math", "English", "Python"]].cov().round(2))
print(df[["Math", "English", "Python"]].corr().round(2))
#Math和Python的协方差是正的，说明两科成绩同向变化；相关系数0.88，正相关很强
```
