```python
import os, uuid, hashlib, platform
# ===== 1. 填写学号 =====
STUDENT_ID = "20244081128"
NAME = "唐仕江"
# ===== 2. 打印 conda 环境 =====
# 注意：CONDA_DEFAULT_ENV 是 conda activate 时注入的变量，
# Jupyter 内核不经过 activate，因此为空时回退为 "base"
CONDA_ENV = os.environ.get("CONDA_DEFAULT_ENV") or "base"
print(f"[Conda 环境] {CONDA_ENV}")
# ===== 3. 机器指纹（绑定当前电脑）=====
MACHINE_FP = hashlib.sha256(
    f"{platform.node()}|{platform.machine()}|{platform.system()}|{uuid.getnode()}".encode()
).hexdigest()[:16]
# ===== 4. 生成唯一 ID =====
RUN_ID = f"{STUDENT_ID}-{MACHINE_FP}"
print(f"[学号]     {STUDENT_ID}")
print(f"[姓名]     {NAME}")
print(f"[机器指纹] {MACHINE_FP}")
print(f"[RUN_ID]   {RUN_ID}")
```

```python
#正则表达式的基础匹配
import re

#任务1
t1 = "cat on mat; later: cats and dogs"
print(re.findall(r"\d+", t1))
print(re.findall(r"\s+", t1))
print(re.findall(r"\bcat\b", t1))
#\d+匹配数字，这句话里没有数字所以是空；\s+匹配空白；\bcat\b只匹配独立的cat，cats不算
print()

#任务2
t2 = "cat dog bird CAT"
print(re.findall(r"cat|dog", t2))
print(re.findall(r"a|aa|aaa", "aaaa"))
print(re.findall(r"aaa|aa|a", "aaaa"))
#交替匹配从左往右试，写在前面的优先，所以想匹配到aaa要把长的写在前面
print()

#任务3
t3 = "ct cat caat caaat"
print(re.findall(r"ca*t", t3))
print(re.findall(r"ca+t", t3))
print(re.findall(r"ca?t", t3))
print(re.findall(r"ca{2}t", t3))
print(re.findall(r"ca{1,2}t", t3))
#a*是0个或多个a，a+是至少1个，a?是0个或1个，a{2}是正好2个，a{1,2}是1到2个
print()
```

```python
#分组、回溯引用与替换
import re

#任务1
t1 = "联系：alice@example.com；备选：bob@test.org"
m1 = re.findall(r"(\w+)@([\w.]+)", t1)
print(m1)
#每个元组第一个是用户名，第二个是域名
print()

#任务2
t2 = "cat cat, dog cat, bird bird"
print(re.findall(r"\b(\w+)\s+\1\b", t2))
#\1是后向引用，匹配和前面捕获的同一个词，找到连续重复的cat和bird
print()

#任务3
t3 = "2024-05-01\n2024/05/02\n2024.05.03"
print(t3)
t4 = re.sub(r"(\d{4})[-/.](\d{2})[-/.](\d{2})", r"\2-\3-\1", t3)
print(t4)
#三个组依次是年月日，替换成月-日-年，行数不变
print()

#任务4
t5 = "联系我：123-456-7890 或 987.654.3210"
t6 = re.sub(r"(\d{3})[-.](\d{3})[-.](\d{4})", r"(\1) \2-\3", t5)
print(t6)
#统一成(123) 456-7890的形式
```

```python
#使用BeautifulSoup解析HTML
from bs4 import BeautifulSoup

html_doc = """<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲：正则表达式</p>
<p class="course">第七讲：结构化数据</p>
<a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a>
</body></html>"""

#任务1
soup = BeautifulSoup(html_doc, "html.parser")
print(soup.title.text)
print(soup.h1.name, soup.h1.text)
print()

#任务2
for p in soup.find_all("p", class_="course"):
    print(p.text)
for a in soup.find_all("a"):
    print(a.text, a["href"])
print()

#任务3
print(soup.get_text())
print(soup.find("p", class_="course").get_text())
print(soup.find("a").get_text())
#整篇的文本包含标题和所有标签内容，单个标签只取自己那段文本
```
