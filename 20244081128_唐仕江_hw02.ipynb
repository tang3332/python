```python
#字符串的乐趣
#任务1
def is_palindrome(s):
    #去掉空格，全部转小写，再比较正序和倒序
    t = s.replace(' ', '').lower()
    return t == t[::-1]

#任务2
def is_abecedarian(s):
    #去掉空格后，两两比较相邻字母，默认转小写
    t = s.replace(' ', '').lower()
    for i in range(len(t) - 1):
        if t[i] > t[i + 1]:
            return False
    return True

#任务3
def remove_vowels(s):
    result = ''
    for c in s:
        if c.lower() not in 'aeiou':
            result += c
    return result

print(is_palindrome("aakkk kkKAa"))
print(is_palindrome("ijjii"))
print(is_abecedarian("asdfghjkl"))
print(is_abecedarian("abdf"))
print(is_abecedarian("13456"))
print(remove_vowels("uhnssbaol"))
```

```python
#列表的乐趣
#任务1
def list_reverse(lst):
    #先检查类型，不是列表就报错
    if not isinstance(lst, list):
        raise TypeError('请输入列表')
    return lst[::-1]

#任务2
def is_sorted(seq):
    for i in range(len(seq) - 1):
        if seq[i] > seq[i + 1]:
            return False
    return True

print(list_reverse([13,432,9,1,0]))
#print(list_reverse("12,45"))
```

```python
#作为向量的元组
#任务1
def vec_scalar_mult(t, s):
    #返回元组t每个元素乘以标量s得到的新元组。
    if isinstance(s, bool) or not isinstance(s, (int, float)):
        raise TypeError("标量s必须是整数或浮点数")
    if not isinstance(t, tuple):
        raise TypeError("向量t必须是元组")
    for i, x in enumerate(t):
        if isinstance(x, bool) or not isinstance(x, (int, float)):
            raise TypeError("t有元素不是整数或浮点数")
    return tuple(x * s for x in t)

#任务2
def vec_inner_product(t1, t2):
    #返回两个等长数字元组的内积（浮点数）
    if not isinstance(t1, tuple):
        raise TypeError("t1必须是元组" )
    if not isinstance(t2, tuple):
        raise TypeError("t2必须是元组")
    if len(t1) != len(t2):
        raise ValueError("两个向量长度必须一致")
    result = 0.0
    for x, y in zip(t1, t2):
        if isinstance(x, bool) or not isinstance(x, (int, float)):
            raise TypeError("t1中有元素不是整数或浮点数")
        if isinstance(y, bool) or not isinstance(y, (int, float)):
            raise TypeError("t2中有元素不是整数或浮点数")
        result += x * y
    return result

#任务3
def check_valid_mx(mx):
    #检查mx是否为合法矩阵（元组的元组，各行等长且元素为数字）
    if not isinstance(mx, tuple) or len(mx) == 0:
        return False
    first = mx[0]
    if not isinstance(first, tuple):
        return False
    for row in mx:
        if not isinstance(row, tuple):        # 外层每个元素都得是元组
            return False
        if len(row) != len(first):            # 所有内层元组长度相同
            return False
        for x in row:
            if isinstance(x, bool) or not isinstance(x, (int, float)):
                return False
    return True

#任务4
def mx_vec_mult(mx, v):
    #返回长度为矩阵行数的元组
    if not check_valid_mx(mx):
        raise TypeError("mx不是合法矩阵（应为元组的元组，各内层元组等长且元素为数字）")
    if not isinstance(v, tuple):
        raise TypeError("v必须是元组")
    if len(mx[0]) != len(v):
        raise ValueError("矩阵列数与向量长度不一致，无法相乘")
    for x in v:
        if isinstance(x, bool) or not isinstance(x, (int, float)):
            raise TypeError("v中有元素不是整数或浮点数")
    return tuple(vec_inner_product(row, v) for row in mx)
```

```python
#更多向量的乐趣
class Vector:

    #任务1
    def __init__(self, dim, elements=None):
        # 维度必须是 int（bool 是 int 子类，先排除），且非负
        if isinstance(dim, bool) or not isinstance(dim, int):
            raise TypeError("维度必须是非负整数")
        if dim < 0:
            raise ValueError("维度不能为负数")
        #只给维度不给元素→默认零向量
        if elements is None:
            elements = (0,) * dim
        if not isinstance(elements, (list, tuple)):
            raise TypeError("elements必须是列表或元组")
        if len(elements) != dim:
            raise ValueError("维度与元素个数不一致")
        for x in elements:
            if isinstance(x, bool) or not isinstance(x, (int, float)):
                raise TypeError("元素中有非整数或浮点数")
        #存成元组，不可变，防止元素被外部改掉后和dim脱节
        self.dim = dim
        self.elements = tuple(elements)

    #任务2
    """
    都必须是实例属性。理由：每个向量对象都有自己独立的维度和元素，这些数据属于“这一个对象”而不是整个类。如果写成类属性，所有Vector实例
    会共享同一份维度和元素，你改一个向量的元素，所有向量都跟着变，向量就不再是独立的数学对象了。类属性适合放“全体实例共享且不变”的东西
    （比如这个类就没有这类需求，顶多放个类文档字符串）
    """
    #内部工具，任务3、4、5共用
    def _check_same_dim(self, other):
        if not isinstance(other, Vector):
            raise TypeError("运算对象必须是元组")
        if self.dim != other.dim:
            raise ValueError("维度不一致，无法比较/运算")

    def _check_scalar(self, s):
        if isinstance(s, bool) or not isinstance(s, (int, float)):
            raise TypeError("标量必须是整数或浮点数")

    #任务3
    #比较运算符（维度不同一律抛异常）
    def __eq__(self, other):
        self._check_same_dim(other)
        return self.elements == other.elements

    def __lt__(self, other):
        self._check_same_dim(other)
        return self.elements < other.elements   #元组比较本身就是字典序

    def __le__(self, other):
        self._check_same_dim(other)
        return self.elements <= other.elements

    def __gt__(self, other):
        self._check_same_dim(other)
        return self.elements > other.elements

    def __ge__(self, other):
        self._check_same_dim(other)
        return self.elements >= other.elements

    def __hash__(self):
        return hash(self.elements)   #定义了__eq__必须补__hash__，否则对象进不了set/dict

    #任务4
    def dot(self, other):
        self._check_same_dim(other)
        return float(sum(a * b for a, b in zip(self.elements, other.elements)))

    #任务5
    def __mul__(self, other):
        if isinstance(other, Vector):
            self._check_same_dim(other)
            return Vector(self.dim, tuple(a * b for a, b in
                                          zip(self.elements, other.elements)))
        self._check_scalar(other)
        return Vector(self.dim, tuple(x * other for x in self.elements))

    __rmul__ = __mul__

    def __repr__(self):
        return "Vector(%d, %r)" % (self.dim, self.elements)

v = Vector(2, (2, 4))
w = Vector(2, (2, 5))
print(Vector(3).elements)
print(v < w)
print(v == Vector(2, (2, 4)))
print(v.dot(w))
print(v * 2)
print(2 * v)
print(Vector(2, (2, 3)) * Vector(2, (4, 5)))
try:
    v < Vector(3, (1, 2, 3))
except ValueError as e:
    print(e)
```
