```python
#函数、条件、循环与递归
#任务1
def score_level(score):
    if not isinstance(score, (int, float)):
        raise ValueError("分数必须是数值")
    if 90 <= score <= 100:
        return "优秀"
    if 80 <= score < 90:
        return "良好"
    if 60 <= score < 80:
        return "及格"
    if 0 <= score < 60:
        return "不及格"
    #以上区间都不命中，说明超出 0~100
    raise ValueError("分数需要在0到100之间")

for v in [95, 85, 72, 40, 0, 100]:
    print(v, "->", score_level(v))
try:
    score_level(120)
except ValueError as e:
    print("捕获到异常：", e)
print()

#任务2
sum_for = 0
for i in range(1,101):
    sum_for += i
sum_while = 0
i = 1
while i <= 100:
    sum_while += i
    i += 1
print("for循环 ",sum_for)
print("while循环 ",sum_while)
#结果是一样的，但是过程不一样，for循环不需要对i进行初始化定义，也不需要i循环加1，while循环需要
print("两者结果相同：", sum_for == sum_while)
print()

#任务3
#迭代版
def fact_iter(n):
    if n < 0:
        raise ValueError("需要输入非负整数")
    if n == 0:
        return 1
    k = 1
    #从2开始，因为k默认是1
    for i in range(2,n+1):
        k *= i
    return k

#递归版
def fact_recur(n):
    if n < 0:
        raise ValueError("需要输入非负整数")
    if n == 0:
        return 1
    return n * fact_recur(n - 1)

for num in [0,1,5,10]:
    print(f"n={num},迭代:{fact_iter(num)},递归：{fact_recur(num)}")
print()

#任务4
def summary(values):
    min_v = min(values)
    max_v = max(values)
    avg_v = sum(values) / len(values)
    return (min_v,max_v,avg_v)

data = [56,62,88,91,54,23]
min_v,max_v,avg_v = summary(data)
print("最小值：",min_v,"，最大值：",max_v,"，平均值：",avg_v)
print()

#任务5
def sum_all(*args):
    sum_t = 0
    for i in args:
        sum_t += i
    return sum_t
print("无参数：",sum_all())
print("单个参数：",sum_all(100))
print("多个参数：",sum_all(1,5,9,3))
print()
```

```python
#文件读写、异常处理与模块
#任务1
import os
import pickle
import copy

cwd = os.getcwd()
print("当前工作目录：",cwd)
data_dir = os.path.join(cwd, "exp01_data")
os.makedirs(data_dir, exist_ok=True)
data_file = os.path.join(data_dir, "students.txt")
print("数据目录：", data_dir)
print("数据文件：", data_file)
print()

#任务2
#用join拼接
raw_text = "\n".join([
    "name,score",
    "Alice,85",
    "Bob,58",
    "Carol,92",
    "David,76",
    "Eva,60",
    "Frank,89",
    "Grace,abc",
    "Henry,105",
    "Incomplete",
]) + "\n"

with open(data_file, "w", encoding="utf-8") as f:
    f.write(raw_text)

with open(data_file, "r", encoding="utf-8") as f:
    lines = f.read().splitlines()

print("共读取行数：", len(lines))
print("表头：", lines[0])
for line in lines[1:]:
    print(repr(line))
print()

#任务3
def parse_student(line):
    """把一行 'name,score' 解析为字典，无效时抛 ValueError。"""
    parts = line.split(",")
    #检查字段数量
    if len(parts) != 2:
        raise ValueError(f"字段数量应为 2，实际为 {len(parts)}")
    name = parts[0].strip()
    score_str = parts[1].strip()
    #检查姓名是否为空
    if not name:
        raise ValueError("姓名为空")
    #成绩必须能转换为整数
    try:
        score = int(score_str)
    except ValueError:
        raise ValueError(f"成绩无法转换为整数：{score_str!r}")
    #成绩必须在 0~100 范围内
    if not (0 <= score <= 100):
        raise ValueError(f"成绩超出 0~100 范围：{score}")
    return {"name": name, "score": score}

#测试三种情况
print(parse_student("Ivy,88"))
for bad in ["Incomplete", "Grace,abc", "Henry,105", ",90"]:
    try:
        parse_student(bad)
    except ValueError as e:
        print(f"{bad!r} -> {e}")
print()

#任务4
#按行处理，捕获异常并继续
valid_records = []
invalid_records = []

#跳过表头
for lineno, line in enumerate(lines[1:], start=2):
    if not line.strip():
        continue
    try:
        valid_records.append(parse_student(line))
    except ValueError as e:
        invalid_records.append({"line": lineno, "content": line, "reason": str(e)})

print(f"有效记录 {len(valid_records)} 条：")
for r in valid_records:
    print("   ", r)
print(f"无效记录 {len(invalid_records)} 条：")
for r in invalid_records:
    print(f"    第{r['line']}行 {r['content']!r} -> {r['reason']}")
print()

#任务5
#把统计结果写入report.txt，并演示w与a模式
n_valid = len(valid_records)
n_invalid = len(invalid_records)
avg_valid = sum(r["score"] for r in valid_records) / n_valid
pass_rate_valid = sum(1 for r in valid_records if r["score"] >= 60) / n_valid

summary_text = "\n".join([
    f"有效记录数：{n_valid}",
    f"无效记录数：{n_invalid}",
    f"平均分：{avg_valid:.2f}",
    f"及格率：{pass_rate_valid:.2%}",
]) + "\n"

report_path = os.path.join(data_dir, "report.txt")

#w模式：不存在就创建，存在则清空后重写
with open(report_path, "w", encoding="utf-8") as f:
    f.write("[第一次 w 写入]\n" + summary_text)
print("--- 第一次 w 写入后 ---")
print(open(report_path, encoding="utf-8").read())

#再次用w写入，原有内容被全部清空
with open(report_path, "w", encoding="utf-8") as f:
    f.write("[第二次 w 写入，原内容已被清空]\n" + summary_text)
print("--- 第二次 w 写入后（注意第一次的内容已消失）---")
print(open(report_path, encoding="utf-8").read())

#a模式：在文件末尾追加，不影响原有内容
with open(report_path, "a", encoding="utf-8") as f:
    f.write("[a 模式追加的一行，原有内容保留]\n")
print("--- a 模式追加后 ---")
print(open(report_path, encoding="utf-8").read())
print("结论：w 会截断旧内容，a 会保留旧内容并在末尾追加")
print()

#任务6
#pickle保存与读取
pkl_path = os.path.join(data_dir, "students.pkl")
with open(pkl_path, "wb") as f:
    pickle.dump(valid_records, f)
print("已保存：", pkl_path)

with open(pkl_path, "rb") as f:
    loaded_records = pickle.load(f)
print("读取到的记录：", loaded_records)
print("与原数据相等：", loaded_records == valid_records)
print()

#任务7
#使用同级目录的student_utils.py模块
import sys
#确保当前目录在模块搜索路径里
if cwd not in sys.path:
    sys.path.insert(0, cwd)

import student_utils

#有效输入
print("有效输入：", student_utils.parse_student("Ivy,88"))
#无效输入
try:
    student_utils.parse_student("Jack,200")
except ValueError as e:
    print("无效输入捕获异常：", e)
print()

#任务8
#读取不存在的文件
missing_path = os.path.join(data_dir, "not_exist.txt")
try:
    with open(missing_path, "r", encoding="utf-8") as f:
        f.read()
except FileNotFoundError as e:
    print("已捕获 FileNotFoundError")
    print("说明：文件不存在，请检查路径是否正确")
    print("系统错误信息：", e)
print()
```

```python
#类、继承与对象复制
#任务1
#Student 类
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        status = "及格" if self.is_passed() else "不及格"
        return f"Student(name={self.name}, score={self.score}, {status})"

#简单验证
print(Student("Test", 75))
print(Student("Test2", 50))
print()

#任务2
#为实验4的每条有效记录创建对象
student_objs = [Student(r["name"], r["score"]) for r in valid_records]
for s in student_objs:
    print(s, "-> is_passed():", s.is_passed())
print()

#任务3
#GraduateStudent继承Student
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        #通过super()初始化父类属性
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        status = "及格" if self.is_passed() else "不及格"
        return (f"GraduateStudent(name={self.name}, score={self.score}, "
                f"topic={self.research_topic}, {status})")

gs = GraduateStudent("Alice", 85, "机器学习")
print(gs)
print("是否继承自 Student：", isinstance(gs, Student))
print("is_passed() 来自父类：", gs.is_passed())
print()

#任务4
#直接赋值/浅拷贝/深拷贝
#每次比较都从新建的原始对象开始
def make_origin():
    return {"name": "Alice", "scores": [85, 90]}

#直接赋值：两个变量指向同一个对象
o1 = make_origin()
a1 = o1
a1["scores"].append(60)
print("原始对象：", o1)
print("赋值后  ：", a1)
#结论：完全同一个对象，改一个另一个也变
print()

#copy.copy()：浅拷贝，外层新对象，内层列表仍是共享的
o2 = make_origin()
a2 = copy.copy(o2)
a2["scores"].append(60)
print("浅拷贝    原始对象：", o2)
print("浅拷贝    副本    ：", a2)
#结论：外层是独立的，但内层scores共享，所以原对象也被改了
print()

#copy.deepcopy()：递归复制，内层列表也是新的
o3 = make_origin()
a3 = copy.deepcopy(o3)
a3["scores"].append(60)
print("深拷贝    原始对象：", o3)
print("深拷贝    副本    ：", a3)
#结论:内层列表也被复制，原对象完全不受影响
print()

#再验证一次外层是否独立
o4 = make_origin()
a4 = copy.copy(o4)
a4["name"] = "Bob"
print("浅拷贝改顶层键后，原对象 name：", o4["name"], "，副本 name：", a4["name"])
print()
```
